****FILTERING THE DATA****

```python
ipl.head()
```

```python
#find all the final answers
ipl[ipl['MatchNumber']=='final'][['Season','WinningTeam']]
```

```python
#how many super over finsihes have occoured 
ipl[ipl['SuperOver']=='Y'].shape[0]
```

```python
#how many mathces csk win in kolkata 
ipl[(ipl['City']=='Kolkata') & (ipl['WinningTeam']=='Chennai Super Kings')].shape[0]
```

```python
#toss winner is winner percentage
(ipl[ipl['TossWinner']==ipl['WinningTeam']].shape[0]/ipl.shape[0])*100
```

```python
#movies with rating higher then 8 and votes >8000
movies[(movies['imdb_rating']>8.3) & (movies['imdb_votes']>10000)]
```

```python
#action movies with rating more than 7
movies.sample(2)
mask=movies['genres'].str.split('|').apply(lambda x:'Action' in x)
mask2=movies['imdb_rating']>7.5
movies[mask & mask2]
```
