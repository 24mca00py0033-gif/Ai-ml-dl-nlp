```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

#### Relational Plots


```python
#scatterplot -> axis level function
tips=sns.load_dataset('tips')
tips.head()
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time')
```

```python
#relplot -> figure level->squre shape 
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time')
```

```python
gap=px.data.gapminder()
gap
```

```python
temp_df=gap[gap['country']=='India']
temp_df
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
temp_df=gap[gap['country'].isin(['India','Pakistan','China'])]
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
temp_df=gap[gap['country'].isin(['India','Brazil','China','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent')
```

```python
#facetplot-> work with relplot not with axis level function
sns.relplot(kind='scatter',data=tips,x='total_bill',y='tip',col='sex',row='day')
```

```python
sns.relplot(data=gap,x='lifeExp',kind='scatter',y='gdpPercap',col='year',col_wrap=3)
```
