## Bar chart

**Use cases**
- Bivariate anayliss
- Nummerical vs categorical
- aggregate anylisis of groups 

```python
#simple graph
childern =[10,20,30,40,50]
colors=['red','green','pink','blue','yellow']
plt.bar(colors,childern,color='red')
```

```python
data=pd.read_csv('batsman_season_record.csv')
```

```python
data
```

```python
plt.bar(data['batsman'],data['2015'])
plt.bar(data['batsman'],data['2016'])
```

```python
plt.bar(np.arange(data.shape[0])-.2,data['2015'],width=0.2)
plt.bar(np.arange(data.shape[0]),data['2016'],width=0.2)
plt.bar(np.arange(data.shape[0])+.2,data['2017'],width=0.2)

plt.xticks(np.arange(data.shape[0]),data['batsman'])
plt.show()
```

```python
#stacked bar chart
data
```

```python
plt.bar(data['batsman'],data['2015'],label='2015')
plt.bar(data['batsman'],data['2016'],bottom=data['2015'],label='2016')
plt.bar(data['batsman'],data['2017'],bottom=data['2016']+data['2015'],label='2017')

plt.legend()
plt.show()


```
