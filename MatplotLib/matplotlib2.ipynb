```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('iris.csv')
data.head()
```

```python
plt.scatter(data['SepalLengthCm'],data['PetalLengthCm'])
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
```

```python
data.sample(5)
data['Species'] = data['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
```

```python
data.sample(5)
```

```python
plt.scatter(data['SepalLengthCm'],data['PetalLengthCm'],c=data['Species'],cmap='jet',alpha=0.5)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.colorbar()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(data['SepalLengthCm'],data['PetalLengthCm'],c=data['Species'],cmap='jet',alpha=0.5)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.colorbar()
```

```python
batter=pd.read_csv('batter.csv')
batter.sample(5)
```

```python
batter.shape
```

```python
sample_df=batter.head(100).sample(25,random_state=5)
sample_df
```

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],sample_df['runs'])

plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(30,color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

### SubPlots

```python
fig,ax=plt.subplots()

ax.scatter(batter['avg'],batter['strike_rate'])
ax.set_title('SubPlot')
ax.set_xlabel('abg')
ax.set_xlabel('avg')
ax.set_ylabel('strike rate')
```

```python
batter.head()
```

```python
fig, ax = plt.subplots(nrows=2, ncols=2)

ax[0][0].scatter(batter['avg'], batter['runs'])

ax[0][1].scatter(batter['avg'], batter['strike_rate'])

ax[1][0].hist(batter['strike_rate'])

ax[1][1].hist(batter['runs'], bins=10)
```

```python
fig=plt.figure()

ax1=fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='green')

ax2=fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])
```

#### 3d Scatter plot

```python
batter
fig=plt.figure()

ax=plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])
```

## HeatMap

```python
delivery=pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.sample(3)
```

```python
temp_df=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run']==6)]
temp_df
```

```python
grid=temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
plt.figure(figsize=(12,8))
plt.imshow(grid)
plt.colorbar()
```
